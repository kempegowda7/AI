```python
import collections

class EightPuzzle:
    def __init__(self, initial_state, final_state):
        self.initial_state = initial_state
        self.final_state = final_state

    # Helper function to find the position of the blank (0) tile
    def find_blank(self, state):
        return state.index(0)

    # Helper function to generate possible next states from a given state
    def get_neighbors(self, state):
        neighbors = []
        blank_idx = self.find_blank(state)

        # Define possible moves based on the blank's position
        # (row, col) coordinates for a 3x3 grid
        row, col = blank_idx // 3, blank_idx % 3

        # Possible directions: (dr, dc)
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)] # Up, Down, Left, Right

        for dr, dc in moves:
            new_row, new_col = row + dr, col + dc

            if 0 <= new_row < 3 and 0 <= new_col < 3:
                # Calculate the index of the tile to swap with the blank
                swap_idx = new_row * 3 + new_col

                # Convert state tuple to list to allow swapping
                new_state_list = list(state)
                new_state_list[blank_idx], new_state_list[swap_idx] = new_state_list[swap_idx], new_state_list[blank_idx]
                neighbors.append(tuple(new_state_list))
        return neighbors

    def dfs(self, depth_limit=None):
        stack = collections.deque([(self.initial_state, [self.initial_state])])
        # Keep track of visited states to avoid cycles and redundant paths within a branch.
        # For finding a 'minimum depth' with pure DFS, Iterative Deepening DFS (IDDFS) is typically used.
        # This implementation performs a Depth-Limited Search.
        visited = {self.initial_state}

        while stack:
            current_state, path = stack.pop()

            if current_state == self.final_state:
                return path

            # Prune this path if it exceeds the depth limit
            if depth_limit is not None and len(path) > depth_limit:
                continue

            # Explore neighbors in reverse order to simulate typical DFS exploration if order matters
            # (e.g., exploring Up, Down, Left, Right neighbors consistently).
            for neighbor in reversed(self.get_neighbors(current_state)):
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.append((neighbor, path + [neighbor]))
        return None # No solution found within the specified depth limit


# Define the initial and final states
initial_state = (2, 8, 3, 1, 6, 4, 7, 0, 5) # 0 represents the blank space
final_state = (1, 2, 3, 8, 0, 4, 7, 6, 5)

print(f"Initial State: {initial_state}")
print(f"Final State: {final_state}")

puzzle = EightPuzzle(initial_state, final_state)

# Run DFS with a depth limit of 7
depth_limit = 7
solution_path = puzzle.dfs(depth_limit=depth_limit)

if solution_path:
    print(f"\nSolution Found within depth limit {depth_limit}!")
    print(f"Solution path length: {len(solution_path) - 1} moves")
    for i, state in enumerate(solution_path):
        print(f"Step {i}: {state[0:3]}\n       {state[3:6]}\n       {state[6:9]}\n")
else:
    print(f"\nNo solution found within depth limit {depth_limit} using DFS for this puzzle configuration.")

# Another example for a state that might not be solvable or takes too long for DFS
# initial_state_hard = (8, 6, 7, 2, 5, 4, 3, 0, 1)
# final_state_easy = (1, 2, 3, 4, 5, 6, 7, 8, 0)
# puzzle_hard = EightPuzzle(initial_state_hard, final_state_easy)
# solution_hard = puzzle_hard.dfs()
# if solution_hard:
#     print(f"Solution path length for hard puzzle: {len(solution_hard)}")
# else:
#     print("No solution found for hard puzzle.")
```
